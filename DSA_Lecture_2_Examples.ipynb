```python
#DSA Lecture 2 Example 1
# List of sales in different regions
sales = [420, 455, 610, 300, 175]

# List of region names
regions = ["North", "South", "East", "West", "Central"]

# Calculate total sales
total = sum(sales)

# Calculate average sales
average = total / len(sales)

# Find the lowest sales value
lowest_sales = min(sales)

# Find the index of the lowest sales
lowest_index = sales.index(lowest_sales)

# Print total sales
print("Total Sales:", total)

# Print average sales
print("Average Sales:", average)

# Print the lowest performing region
print("Lowest Performing Region:", regions[lowest_index])
```

```python
#DSA Lecture 2 Example 2
# Import the matplotlib library for creating graphs
import matplotlib.pyplot as plt

# List of regions
regions = ["North", "South", "East", "West", "Central"]

# Sales data for each region
sales = [420, 455, 610, 300, 175]

# Create a bar chart
plt.bar(regions, sales)

# Add a title to the graph
plt.title("Regional Sales")

# Add a label to the Y-axis
plt.ylabel("Sales (₹ thousands)")

# Display the graph
plt.show()
```

```python
#DSA Lecture 2 Example 3
# List of regions
regions = ["North", "South", "East", "West", "Central"]

# Sales data for each region
sales = [420, 455, 610, 300, 175]

# Calculate the average sales of all regions except Central
avg_others = sum(s for r, s in zip(regions, sales) if r != "Central") / 4

# Find the sales gap between the average and Central region
gap = avg_others - sales[regions.index("Central")]

# Print the result
print("Central trails the other regions' average by", round(gap, 1), "(₹ thousands)")
```
