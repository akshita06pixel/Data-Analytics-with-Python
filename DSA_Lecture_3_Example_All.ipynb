```python
# Store an integer value
a = 10

# Store a decimal value
b = 10.5

# Store a text value
c = "Data Analysis"

# Store a True/False value
d = True

# Print the data type of each variable
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

```python
# Store 100 as a text (string)
x = "100"

# Convert the string into an integer
# Then add 50 to it
y = int(x) + 50

# Print the result and its data type
print(y)
print(type(y))
```

```python
# Create a list of fruits
fruits = ["apple", "banana", "mango"]

# Check if "banana" is present in the list
print("banana" in fruits)
```

```python
# Create a tuple with three numbers
t = (1, 2, 3)

# Print the tuple and its data type
print(t)
print(type(t))

# Try to change the first value of the tuple
try:
    t[0] = 99

# A tuple cannot be changed, so it gives a TypeError
except TypeError as e:
    print("Error:", e)
```

```python
# Create a set with some numbers
s = {1, 2, 2, 3}

# Print the set
print(s)
```

```python
# Create a dictionary with two values
record = {
    "a": 1,
    "b": 2
}

# Access and print the value of key "a"
print(record["a"])
```

```python
# Create a complex number
z = 3 + 4j

# Print the real part of the complex number
print("Real part:", z.real)

# Print the imaginary part of the complex number
print("Imaginary part:", z.imag)

# Print the magnitude of the complex number
print("Magnitude:", abs(z))
```
