```python
# Create a list of different types of data samples
samples = [
    {
        "type": "structured",
        "value": {"Name": "Raj", "Age": 20}
    },
    {
        "type": "semi-structured",
        "value": {"tags": ["python", "data"]}
    },
    {
        "type": "unstructured",
        "value": "A long free-text customer comment..."
    }
]

# Loop through each sample
for s in samples:

    # Print the type of data and the data type of its value
    print(f"{s['type']}: {type(s['value']).__name__}")
```
